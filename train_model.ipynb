```python
import pandas as pd
import pickle

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
df = pd.read_csv(r"C:\Users\admin\Desktop\PythonProject\PythonProject1\PythonProject\PythonProject\PythonProject\ML_streamlit\INSURANCE (2).csv")

print("Dataset loaded successfully")

print(df.head())
```

```python
X = df.drop("charges", axis=1)

y = df["charges"]
```

```python
categorical_columns = [
    "sex",
    "smoker",
    "region"
]

numerical_columns = [
    "age",
    "bmi",
    "children"
]
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_columns
        )
    ],
    remainder="passthrough"
)
```

```python
model = LinearRegression()
```

```python
pipeline = Pipeline(
    steps=[
        ("preprocessing", preprocessor),
        ("model", model)
    ]
)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
pipeline.fit(X_train, y_train)

print("Model training completed!")
```

```python
y_pred = pipeline.predict(X_test)
```

```python
mae = mean_absolute_error(y_test, y_pred)
```

```python
mse = mean_squared_error(y_test, y_pred)
```

```python
rmse = mse ** 0.5
```

```python
r2 = r2_score(y_test, y_pred)
```

```python
print("\n========== MODEL PERFORMANCE ==========")

print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R2  :", r2)
```

```python
import pickle

with open("model/insurance_model.pkl", "wb") as file:
    pickle.dump(pipeline, file)

print("Model saved successfully!")
print("File: model/insurance_model.pkl")
```
